# exceptions

> The exception hierarchy for `morphon_nbdev`. Every error raised by this package
> inherits from `GeomorphologyError`, so callers can catch one base type instead of
> guessing which specific error a given function might raise.

```python
#| default_exp exceptions
```

### The hierarchy

- `GeomorphologyError` -- base class; catch this to handle any error from this package generically.
- `UnprojectedCRSError` -- raised by `load_bathymetry` when the input raster has no CRS, or a geographic (lat/lon) one instead of a projected one.
- `InvalidParameterError` -- raised when a function argument is unusable given the other arguments (e.g. `show_map` called with nothing to plot), as opposed to simply the wrong type (which Python's own `TypeError` already covers).

```python
#| export
class GeomorphologyError(Exception):
    "Base exception for all morphon_nbdev errors."

class UnprojectedCRSError(GeomorphologyError):
    "The input raster has no CRS, or a geographic (lat/lon) one; a projected CRS is required."

class InvalidParameterError(GeomorphologyError):
    "A function argument is the wrong type, out of range, or otherwise unusable, e.g. show_map called with nothing to plot."
```
