# morphon

> Self-implemented seabed geomorphology classification (Dove/Nanson two-part MIM-GA scheme) for Bass Strait -- Part 1 Morphology and Part 2 Geomorphology.

## What this is

A ground-up (`numpy`/`scipy`/`xarray`/`rioxarray`/`geopandas`/`shapely`/`rasterio`/
`networkx`/`scikit-image`, `hvplot`/`holoviews` for visualisation) implementation of
Dove et al.'s Part 1 Morphology glossary and Nanson et al.'s Part 2 Geomorphology
framework -- deliberately without WhiteboxTools, RichDEM, GRASS, or SAGA, so every
algorithm is understood and owned rather than called into as a black box.

This is a separate, fresh build from the [`geomorphology_tools`](../geomorphology_tools)
project (a direct port of Geoscience Australia's GA-SaMMT), not a continuation of it.

## The pipeline

1. **Terrain derivatives** -- slope, aspect, curvature, TPI/BPI, Openness,
   Geomorphons, Wood's landform classification.
2. **Boundary refinement** -- LMI/CI hybrids extending TPI's core into full polygons.
3. **Polygonization** -- raster classification to vector features.
4. **Lineament extraction** -- Crest, Thalweg, Centreline, Break in slope, Sill.
5. **Attribute calculation** -- shape, topographic, and profile attributes per polygon.
6. **Classification** -- assigning each polygon one of the 40 Part 1 Morphology
   terms (fuzzy/weight-of-evidence rules, not hard thresholds).
7. **Part 2** -- Setting/Process, BGU, BGU-T, BGU-sT geomorphology classification,
   building on the Part 1 output.

Full algorithm-level detail (formulas, citations, per-term discriminators) lives in
`~/geomorphology_algorithm_reference.md`, kept outside this repo since it's a
standalone research reference, not project code.

## Development install

Research code for an active PhD project, not published to PyPI/conda. Uses
[`uv`](https://docs.astral.sh/uv/):

```sh
cd morphon
uv sync
```

Then either run the notebooks under `nbs/` directly (`uv run jupyter lab`), or use
the compiled package:

```python
from morphon_nbdev.core import load_bathymetry
```

Changed something under `nbs/`? Recompile with `uv run nbdev-export`, then check it
with `uv run nbdev-test`.
